```python
# taken from lab 3
def euler_newton(force, times, x0, v0, m,dt):
    # create arrays for x and v
    x = []
    v = []
    # store the initial conditions in the first elements
    x.append(x0)
    v.append(v0)
    
    for i in range(nsteps-1):
        a = force(times[i], x[i], v[i]) / m
        v.append(v[i] + (a * dt))
        x.append(x[i] + (v[i] * dt))
    return x, v
nsteps=1000
times,dt = np.linspace(0, 10, nsteps, retstep=True)
```

```python
# terminal v for lin. drag
# terminal v means a=0
g = 9.8
b=10
m=10
# V = -(m / b) * g
def lin_term(t, x, v):
    F=-m*g-b*v
    return F
print(lin_term(10, 10, 10))
x0=10
v0=0
x,v=euler_newton(lin_term, times, x0, v0, m,dt)
```

```python
# x vs t
plt.plot(times,x)
plt.grid
plt.title("Position vs Time for Linear Drag")
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.tight_layout()
plt.show()
# v vs t
plt.plot(times,v)
plt.grid
plt.title("Velocity vs Time for Linear Drag")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.tight_layout()
x = [0,10]
y = [-(m / b)*g,-(m / b)*g]
plt.plot(x,y)
plt.show()
```

```python
# terminal v for quad. drag
c=10
def quad_term(t, x, v):
    G = -m*g-c*v*abs(v)
    return G
print(quad_term(10,10,10))
x0=0
v0=0
x,v=euler_newton(quad_term, times, x0, v0, m,dt)

# x vs t
plt.plot(times,x)
plt.grid
plt.title("Position vs Time for Quadratic Drag")
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.tight_layout()
plt.show()
# v vs t
plt.plot(times,v)
plt.grid
plt.title("Velocity vs Time for Quadratic Drag")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.tight_layout()
x = [0,10]
y = [-np.sqrt(m*g/c),-np.sqrt(m*g/c)]
plt.plot(x,y)
plt.show()
```

Linear Drag decreases the velocity on a smoother scale than the drop off from quadratic drag. Quadratic drag reaches terminal velocity faster than linear drag. Terminal velocity for linear drag is v(t)=mg/b and the terminal velocity for quadratic drag is v(t)=sqrt(mg/c). t, m, x, b, c are dummy variables.
